```python
from pyspark.sql.functions import split

# Properly parse movies dataset (delimited by ::)
raw_movies = spark.read.text("/movies.dat")
df_movies = raw_movies.select(
    split(raw_movies.value, "::").getItem(0).alias("movie_id"),
    split(raw_movies.value, "::").getItem(1).alias("movie_name"),
    split(raw_movies.value, "::").getItem(2).alias("movie_type")
)

# Properly parse ratings dataset (delimited by ::)
raw_ratings = spark.read.text("/ratings.dat")
df_ratings = raw_ratings.select(
    split(raw_ratings.value, "::").getItem(0).alias("movie_id"),
    split(raw_ratings.value, "::").getItem(1).alias("user_id"),
    split(raw_ratings.value, "::").getItem(2).alias("rating"),
    split(raw_ratings.value, "::").getItem(3).alias("timestamp")
)

# Perform the join on the correctly parsed 'movie_id' column
df_movies_ratings = df_movies.join(df_ratings, "movie_id", "inner")

# Display the structured result
df_movies_ratings.show()
```

```python
from pyspark.sql.functions import split

# 1. Properly parse users dataset (delimited by ::) with meaningful column names
raw_users = spark.read.text("/users.dat")
df_users = raw_users.select(
    split(raw_users.value, "::").getItem(0).alias("user_id"),
    split(raw_users.value, "::").getItem(1).alias("gender"),
    split(raw_users.value, "::").getItem(2).alias("age"),
    split(raw_users.value, "::").getItem(3).alias("occupation"),
    split(raw_users.value, "::").getItem(4).alias("zip_code")
)

# 2. Join the three tables (df_movies_ratings is already movies + ratings joined on 'movie_id')
# Now, we join that result with our structured df_users on 'user_id'
df_all_joined = df_movies_ratings.join(df_users, "user_id", "inner")

# 3. Display the final merged result
df_all_joined.show()
```
